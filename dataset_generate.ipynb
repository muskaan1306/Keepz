```python
for i in range(NUM_EMPLOYEES):

    employee = {}

    # EMPLOYEE
    employee["emp_id"] = f"E_{i+1:05d}"
    employee["name"] = fake.name()
    employee["location"] = random.choices(list(location_weights.keys()),
        weights=location_weights.values(),
        k=1
    )[0]
    employee["latitude"], employee["longitude"] = hubs[employee["location"]]
    employee["department"] = random.choice(departments)
    employee["role"] = random.choice(roles_by_department[employee["department"]])
    employee["job_level"] = role_level_map[employee["role"]]
    employee["gender"] =fake.passport_gender()
    employee["age"] = random.randint(22, 60)

    #CAREER
    employee["tenure_years"] = round(np.random.gamma(2, 2),1)
    employee["position_tenure"] = round(random.uniform(0.5,employee["tenure_years"]),1)
    employee["time_since_joining_team"] = round(random.uniform(0,min(5, employee["tenure_years"])),1)  
    employee["promotion_delay_months"] = random.choices([6,12,18,24,36,48,60],weights=[10,20,25,25,12,5,3])[0] 
    employee["internal_job_applications"] = np.random.poisson(1)

    #ENGAGEMENT
    employee["manager_relationship"] = round(random.uniform(1,5),1)
    engagement = (employee["manager_relationship"]*0.5 + random.uniform(1,3))
    employee["engagement"] = round(min(5,max(1,engagement)),1)
    employee["work_life_balance"] = round(random.uniform(1,5),1)
employee["peer_recognition_count"] = np.random.poisson(4)
employee["awards_received"] = random.choices(
    [0,1,2,3],
    weights=[70,20,8,2]
)[0]


#WORKLOAD
employee["workload"] = random.randint(1,5)
employee["overtime_hours"] = random.randint(
    employee["workload"] * 4,
    employee["workload"] * 12
)
employee["average_weekly_hours"] = (
    40 +
    employee["overtime_hours"]/4
)
employee["client_escalation_count"] = np.random.poisson(
    employee["workload"]
)

#COMPENSATION
salary_ranges = {
    1:(25000,50000),
    2:(50000,120000),
    3:(120000,220000),
    4:(220000,350000),
    5:(350000,600000),
    6:(600000,1200000)
}
employee["salary_percentile"] = random.randint(1,100)
employee["compensation_gap"] = round(
    random.uniform(-25,25),
    1
)

#MOBILITY
employee["distance_to_office_km"] = round(
    np.random.gamma(2,8),
    1
)
employee["work_flexibility_score"] = random.randint(1,5)
employee["remote_work_eligibility"] = random.choices(
    [0,1],
    weights=[40,60]
)[0]

#MARKET DEMAND
Cloud Engineer = 5
AI Engineer = 5
Data Scientist = 5
DevOps Engineer = 5
Software Engineer = 4
Support Engineer = 2
HR Executive = 2

employee["skill_criticality_score"] = random.randint(1,5)
employee["knowledge_risk_score"] = min(
    5,
    int(
        employee["job_level"] +
        random.randint(-1,1)
    )
)
employee["dependencies"] = max(
    0,
    int(
        np.random.normal(
            employee["job_level"] * 3,
            2
        )
    )
)
    employees.append(employee)
```
